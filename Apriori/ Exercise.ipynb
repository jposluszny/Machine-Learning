```python
import pandas as pd
from efficient_apriori import apriori
```

```python
df = pd.read_csv('Online Retail.csv').sample(1000)
```

```python
df.isna().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isna().sum().sum()
```

```python
df.info()
```

```python
df.describe()
```

```python
transactions = df.groupby('InvoiceNo')['Description'].apply(list).to_list()
```

```python
itemsets, rules = apriori(transactions=transactions, min_support=0.001, min_confidence=0.3)
```

```python
rules_date = []
for rule in rules:
    rules_date.append({'Rule': ', '.join(rule.lhs) + ' --> ' + ', '.join(rule.rhs), 'Lift' : rule.lift})
```

```python
results = pd.DataFrame(rules_date).sort_values('Lift', ascending=False)
```

```python
results.head()
```

```python
results.to_csv('effiecient_df.csv', index=False)
```
